### Bisection

**Bisection.** We bracket the root in $[0,1]$ because
$f(0)=-1<0$ and $f(1)=1>0$, so by continuity a root lies inside.

Each iteration halves the interval: compute the midpoint

$$
c = \frac{a+b}{2},
$$

then keep whichever half still contains the sign change.

Convergence is **linear** — roughly one extra binary digit of accuracy per
step — but it is guaranteed as long as the bracket is valid.

We stop when the bracket half-width

$$
\frac{b-a}{2} \le 10^{-12}
$$

or the residual

$$
|f(c)| \le 10^{-12}.
$$

Every midpoint is stored in a `history` array so the path can be plotted.


```julia
using Plots

# Define the function
f(x) = x^3 + x - 1
f_prime(x) = 3x^2 + 1

# True root (for reference)
true_root = 0.6823278038280193

function bisection(f, a, b; tol=1e-12)
    history = Float64[]
    
    for i in 1:1000
        c = (a + b) / 2
        push!(history, c)
        
        if abs(b - a) / 2 <= tol || abs(f(c)) <= tol
            return c, history
        end
        
        if f(a) * f(c) < 0
            b = c
        else
            a = c
        end
    end
    
    return (a + b) / 2, history
end

root_bis, hist_bis = bisection(f, 0.0, 1.0)
```

### Newton

**Newton.** Starting from $x_0=0.5$, we follow the tangent line at the
current iterate:

$$
x_{n+1}
=
x_n-\frac{f(x_n)}{f'(x_n)}.
$$

Near a simple root with $f'(r)\neq0$, convergence is **quadratic** — the
number of correct digits roughly doubles each iteration, which is why only a
handful of steps are needed here.

The trade-off is that an analytic derivative

$$
f'(x)=3x^2+1
$$

is required, and a poor starting point can cause divergence or cycling.

We stop when the step size

$$
|x_{n+1}-x_n|\le10^{-12}
$$

or the residual

$$
|f(x_{n+1})|\le10^{-12}.
$$

```julia
function newton(f, f_prime, x0; tol=1e-12)
    history = Float64[x0]
    x = x0
    
    for i in 1:1000
        x_new = x - f(x) / f_prime(x)
        push!(history, x_new)
        
        if abs(x_new - x) <= tol || abs(f(x_new)) <= tol
            return x_new, history
        end
        
        x = x_new
    end
    
    return x, history
end
root_new, hist_new = newton(f, f_prime, 0.5)
```

```julia
function secant(f, x0, x1; tol=1e-12)
    history = Float64[x0, x1]
    x_prev = x0
    x_curr = x1
    
    for i in 1:1000
        x_next = x_curr - f(x_curr) * (x_curr - x_prev) / (f(x_curr) - f(x_prev))
        push!(history, x_next)
        
        if abs(x_next - x_curr) <= tol || abs(f(x_next)) <= tol
            return x_next, history
        end
        
        x_prev = x_curr
        x_curr = x_next
    end
    
    return x_curr, history
end
root_sec, hist_sec = secant(f, 0.0, 1.0)
```

### Secant

**Secant.** Same update as Newton, but the derivative is replaced by a
finite-difference estimate built from the two most recent iterates:

$$
x_{n+1}
=
x_n
-
f(x_n)
\frac{x_n-x_{n-1}}
     {f(x_n)-f(x_{n-1})}.
$$

We start from the two seeds

$$
x_0=0,
\qquad
x_1=1.
$$

No analytic derivative is needed, but two starting points are required.

Convergence is **superlinear**, with order equal to the golden ratio

$$
p=\frac{1+\sqrt{5}}{2}\approx1.618,
$$

which is between bisection and Newton.

We stop when

$$
|x_{n+1}-x_n|\le10^{-12}
$$

or

$$
|f(x_{n+1})|\le10^{-12},
$$

and we guard against a zero denominator before each update.

```julia
using Printf, Plots

# Constants
hbar = 1.054571817e-34
m = 9.1093837015e-31
eV = 1.602176634e-19

a = 0.5e-9
V0 = 50eV
λ = a*sqrt(2*m*V0)/hbar

println("lambda = ", λ)

# Eigenvalue equations
fe(z) = z*tan(z) - sqrt(λ^2-z^2)
fo(z) = -z/tan(z) - sqrt(λ^2-z^2)

# Bisection
function bisect(f,L,R,ϵ=1e-10)
    fL = f(L)
    fL*f(R) < 0 || error("No sign change")
    zhist = Float64[]
    fhist = Float64[]
    
    while (R-L)/2 > ϵ
        z = (L+R)/2
        fz = f(z)

        push!(zhist,z)
        push!(fhist,abs(fz))
        abs(fz) <= ϵ && break
        if fL*fz < 0
            R = z
        else
            L = z
            fL = fz
        end
    end
    return (L+R)/2,zhist,fhist
end

# Find roots
δ = 1e-12
even = Float64[]
odd = Float64[]

even_z = Float64[]
even_f = Float64[]
odd_z = Float64[]
odd_f = Float64[]

n = 0

while n*pi < λ

    # Even interval
    L = n*pi + δ
    R = min(n*pi + pi/2,λ) - δ

    if L < R && fe(L)*fe(R) < 0
        root,zs,fs = bisect(fe,L,R)
        push!(even,root)
        if isempty(even_z)
            even_z = zs
            even_f = fs
        end
    end

    # Odd interval
    L = n*pi + pi/2 + δ
    R = min((n+1)*pi,λ) - δ

    if L < R && fo(L)*fo(R) < 0
        root,zs,fs = bisect(fo,L,R)
        push!(odd,root)

        if isempty(odd_z)
            odd_z = zs
            odd_f = fs
        end
    end
    n += 1
end

# ------------------------------------------------------------
# Print convergence side-by-side
# ------------------------------------------------------------

println("\nBisection convergence")
println()

@printf("%4s %12s %18s %18s %18s\n", "k", "z_even",
    "|f_even|", "z_odd", "|f_odd|")

println("-"^82)

N = max(length(even_z),length(odd_z))

for k in 1:N

    if k <= length(even_z)
        ze = @sprintf("%.12f",even_z[k])
        fev = @sprintf("%.10e",even_f[k])
    else
        ze = ""
        fev = ""
    end

    if k <= length(odd_z)
        zo = @sprintf("%.12f",odd_z[k])
        fod = @sprintf("%.10e",odd_f[k])
    else
        zo = ""
        fod = ""
    end

    @printf(
        "%4d %18s %18s %18s %18s\n",
        k,ze,fev,zo,fod
    )
end

# ------------------------------------------------------------
# Energies
# ------------------------------------------------------------

energy(z) = (-V0 + hbar^2*z^2/(2*m*a^2))/eV
states = [(z,energy(z),"even") for z in even]
append!(states,[(z,energy(z),"odd") for z in odd])
sort!(states,by=x->x[2])


println("\nBound states:")
@printf("%3s %8s %10s %14s\n","n","Parity","z","E (eV)")

for (i,(z,E,p)) in enumerate(states)
    @printf("%3d %8s %14.10f %14.8f\n", i-1,p,z,E)
end

# ============================================================
# Plots
# ============================================================

gr()
default(fontfamily="Computer Modern", foreground_color=:black, background_color=:white,
    grid=false, framestyle=:box, tick_direction=:in, linewidth=1.2,
    dpi=400, size=(600,400))

z = range(0.01,5pi,length=5000)

ytan = [abs(cos(x)) < 0.02 ? NaN : tan(x) for x in z]
ycot = [abs(sin(x)) < 0.02 ? NaN : 1/tan(x) for x in z]

ticks = 0:pi/2:5pi
labels = ["0","π/2","π","3π/2","2π","5π/2","3π","7π/2","4π","9π/2","5π"]

# tan(z)
p1 = plot(z,ytan, color=:black, linewidth=1.2, xlabel="z", ylabel="tan(z)",
    xlims=(0,5pi), ylims=(-8,8), xticks=(ticks,labels), legend=false)

for n in 0:4 vline!(p1,[(n+0.5)*pi], color=:red,
        linestyle=:dash, linewidth=0.9)
end

savefig(p1,"tan_z.png")
display(p1)


# cot(z)
p2 = plot(z,ycot, color=:black, linewidth=1.2, xlabel="z", ylabel="cot(z)", 
    xlims=(0,5pi), ylims=(-8,8), xticks=(ticks,labels),legend=false)

for n in 0:5 vline!(p2,[n*pi], color=:red,
        linestyle=:dash, linewidth=0.9)
end

savefig(p2,"cot_z.png")
display(p2)

# Bisection convergence plot
p3 = plot(1:length(even_f), even_f, yscale=:log10, color=:black, marker=:circle,
    markersize=3, linewidth=1.2, xlabel="Iteration", ylabel="|f(z)|",
    legend=:topright, label="Even state", grid=true)

plot!(p3, 1:length(odd_f), odd_f, color=:red, marker=:square,
    markersize=3, linewidth=1.2, label="Odd state")


savefig(p3,"convergence.png")
display(p3)
```
